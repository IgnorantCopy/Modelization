# 数据预处理

```python
import pandas as pd
import numpy as np
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.express as px
```

```python
data = pd.read_excel("./data/report.xlsx")
data.head()
```

```python
data = data.drop(columns=["序号", "提交答卷时间", "所用时间", "来源", "来源详情", "来自IP", "总分"], axis=1)
data.head()
```

```python
data.columns = ["grade", "gender", "major", "use AI", "usage", "other usage", "efficiency improvement", "question type", "other question type", "frequency1", "frequency2", "helpfulness", "difficulty", "other difficulty", "reasons", "other reasons", "deep mind score", "pressure", "worry", "skills", "other skills", "methods", "other methods"]
data["grade"] = data["grade"].map({"大一": 1, "大二": 2, "大三": 3, "大四及以上": 4})
data["gender"] = data["gender"].map({"男": 0, "女": 1})
data["major"] = data["major"].map({"理工科": 0, "文科": 1, "商科": 2, "医科": 3})
data["use AI"] = data["use AI"].map({"有": 1, "无": 0})
use_data = data[data["use AI"] == 1]
use_data.reset_index(inplace=True)
not_use_data = data[data["use AI"] == 0]
not_use_data.reset_index(inplace=True)
use_data.head()
```

```python
use_data["efficiency improvement"] = use_data["efficiency improvement"].map({"(跳过)": np.nan, "没有提高反而下降": 0, "提高不明显": 1, "有一定提高": 2, "显著提高": 3})
use_data["frequency1"] = use_data["frequency1"].map({"极少或从不": 0, "每周1次左右": 1, "每周3次以上": 2, "≥每天1次": 3})
use_data["frequency2"] = use_data["frequency2"].map({"极少或从不": 0, "每周1次左右": 1, "每周3次以上": 2, "≥每天1次": 3})
use_data["helpfulness"] = use_data["helpfulness"].map({"没有帮助": 0, "帮助不大": 1, "有一定帮助": 2, "是，有很大帮助": 3})
use_data["difficulty"] = use_data["difficulty"].map({"技术操作复杂，难以掌握": 1, "信息过载，难以筛选有效信息": 2, "依赖AI，导致自己思考减少": 3, "担心数据隐私和安全问题": 4, "其他（请说明）__________": 5})
use_data["pressure"] = use_data["pressure"].map({"没有感到压力，因为我对未来有信心": 0, "压力不大，因为我有足够的学习资源": 1, "有一定压力，但还能应对": 2, "是，感到很大压力": 3})
use_data["worry"] = use_data["worry"].map({"完全不担心": 0, "不太担心，因为AI无法完全替代人类或我有其他竞争优势": 1, "有些担心，但认为还有时间调整": 2, "非常担心，认为可能性很大": 3})
use_data = use_data.drop(columns=["reasons", "other reasons"], axis=1)
use_data.head()
```

```python
usages = ["智能辅导（如AI助教、在线答疑等）", "编程辅助（如代码补全、语法检查等）", "学术问题搜索与引用"]
use_data.insert(5, "智能辅导（如AI助教、在线答疑等）", 0)
use_data.insert(6, "编程辅助（如代码补全、语法检查等）", 0)
use_data.insert(7, "学术问题搜索与引用", 0)
for i, usage, in enumerate(use_data["usage"].tolist()):
    usage = usage.split("┋")
    for u in usage:
        if u in usages:
            use_data.loc[i, u] = 1
use_data = use_data.drop(columns=["usage"], axis=1)
use_data = use_data.rename(columns={"智能辅导（如AI助教、在线答疑等）": "智能辅导", "编程辅助（如代码补全、语法检查等）": "编程辅助", "学术问题搜索与引用": "学术问题搜索与引用"})
use_data.head()
```

```python
questions = ["解题（这道题怎么做？）", "信息检索（什么是…（某种观念等）？）", "启发（我还能做什么？）", "概括或发散内容（帮我写摘要）", "无目的地畅聊（你们AI究竟有没有自我意识？）"]
use_data.insert(11, "解题（这道题怎么做？）", 0)
use_data.insert(12, "信息检索（什么是…（某种观念等）？）", 0)
use_data.insert(13, "启发（我还能做什么？）", 0)
use_data.insert(14, "概括或发散内容（帮我写摘要）", 0)
use_data.insert(15, "无目的地畅聊（你们AI究竟有没有自我意识？）", 0)
for i, question, in enumerate(use_data["question type"].tolist()):
    question = question.split("┋")
    for q in question:
        if q in questions:
            use_data.loc[i, q] = 1
use_data = use_data.drop(columns=["question type"], axis=1)
use_data = use_data.rename(columns={"解题（这道题怎么做？）": "解题", "信息检索（什么是…（某种观念等）？）": "信息检索", "启发（我还能做什么？）": "启发", "概括或发散内容（帮我写摘要）": "概括或发散内容", "无目的地畅聊（你们AI究竟有没有自我意识？）": "无目的地畅聊"})
use_data.head()
```

```python
skills = ["信息筛选与批判性思维能力", "持续学习的能力与意愿", "职业道德与伦理意识", "跨领域合作与沟通能力", "创新与创造力", "技术素养与编程能力"]
use_data.insert(25, "信息筛选与批判性思维能力", 0)
use_data.insert(26, "持续学习的能力与意愿", 0)
use_data.insert(27, "职业道德与伦理意识", 0)
use_data.insert(28, "跨领域合作与沟通能力", 0)
use_data.insert(29, "创新与创造力", 0)
use_data.insert(30, "技术素养与编程能力", 0)
for i, skill, in enumerate(use_data["skills"].tolist()):
    skill = skill.split("┋")
    for s in skill:
        if s in skills:
            use_data.loc[i, s] = 1
use_data = use_data.drop(columns=["skills"], axis=1)
use_data.head()
```

```python
methods = ["参加相关培训课程或讲座", "自主学习新知识、新技能", "参与科研项目或实践活动", "加强与同学、老师的交流与合作", "寻求专业导师或行业专家的指导"]
use_data.insert(32, "参加相关培训课程或讲座", 0)
use_data.insert(33, "自主学习新知识、新技能", 0)
use_data.insert(34, "参与科研项目或实践活动", 0)
use_data.insert(35, "加强与同学、老师的交流与合作", 0)
use_data.insert(36, "寻求专业导师或行业专家的指导", 0)
for i, method, in enumerate(use_data["methods"].tolist()):
    method = method.split("┋")
    for m in method:
        m = m.strip()
        if m in methods:
            use_data.loc[i, m] = 1
use_data = use_data.drop(columns=["methods"], axis=1)
use_data.head()
```

```python
heatmap_data = use_data.drop(columns=["index", "use AI", "other usage", "other question type", "other difficulty", "other skills", "other methods"], axis=1)
heatmap_data = (heatmap_data - heatmap_data.min()) / (heatmap_data.max() - heatmap_data.min())
heatmap_data.head()
```
