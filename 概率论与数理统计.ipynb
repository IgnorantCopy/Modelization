# 概率论与数理统计

## 1. scipy.stats 模块

### 1.1 连续性随机变量及其分布
连续型随机变量均有以下方法：
+ rvs: 产生随机样本，通过 size 参数指定样本大小
+ pdf: 概率密度函数
+ cdf: 分布函数
+ sf: 生存函数，即 1-cdf
+ ppf: 分布函数的反函数
+ stat: 计算随机样本的期望和方差
+ fit: 对一组随机样本利用极大似然估计法估计参数

常见的连续型随机变量分布：
+ 均匀分布：uniform.pdf(x, a, b)
+ 指数分布：expon.pdf(x, scale=lambda)
+ 正态分布：norm.pdf(x, mu, sigma)
+ $\chi^2$ 分布：chi2.pdf(x, n)
+ t 分布：t.pdf(x, n)
+ F 分布：f.pdf(x, m, n)
+ $\Gamma$ 分布：gamma.pdf(x, a=A, scale=B)

### 1.2 离散性随机变量及其分布
+ 二项分布：binom.pmf(x, n, p)
+ 泊松分布：poisson.pmf(x, lambda)
+ 几何分布：geom.pmf(x, p)

### 1.3 概率密度函数和分布律的可视化

```python
from scipy.stats import expon, gamma
import plotly.express as px
import numpy as np
import pandas as pd

x = np.linspace(0, 3, 100)
coef = [1 / 3, 1, 2]
temp = []
for i in coef:
    y = expon.pdf(x, scale=i)
    for j, xx in enumerate(x):
        temp.append({"type": "指数分布", 'θ': i, 'x': xx, 'y': y[j]})
for i in coef:
    y = gamma.pdf(x, a=1, scale=i)
    for j, xx in enumerate(x):
        temp.append({"type": "Γ 分布", 'θ': i, 'x': xx, 'y': y[j]})
data = pd.DataFrame(temp)
fig = px.scatter(data, x='x', y='y', color='θ', facet_col='type', labels={'y': "f(x)"})
fig.show()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
from scipy.stats import binom

n = 6
p = 0.3
x = np.arange(n+1)
y = binom.pmf(x, n, p)
fig = go.Figure()
for i in range(n + 1):
    fig.add_trace(go.Line(x=[x[i], x[i]], y=[0, y[i]], line=dict(color="blue"), showlegend=False, mode="lines"))
fig.add_trace(go.Scatter(x=x, y=y, mode="markers", marker=dict(color="blue"), name="二项分布"))
fig.show()
```

## 2. 统计量与统计图

### 2.1 统计量的计算

#### numpy
+ mean: 均值
+ median: 中位数
+ ptp: 极差
+ var: 方差
+ std: 标准差
+ cov: 协方差
+ corrcoef: 相关系数

```python
import numpy as np
import pandas as pd

data = pd.read_csv("./data/data9_5.txt", header=None)
data = data.values   # 转换为 numpy 数组
print(f"均值：{np.mean(data, axis=1)}")
print(f"中位数：{np.median(data, axis=1)}")
print(f"极差：{np.ptp(data, axis=1)}")
print(f"方差：{np.var(data, axis=1, ddof=1)}")     # 无偏方差
print(f"标准差：{np.std(data, axis=1, ddof=1)}")
print(f"协方差：{np.cov(data)}")
print(f"相关系数：{np.corrcoef(data)}")
```

#### pandas
+ describe: 包含均值、标准差、最小值、最大值、四分位数等统计量
+ skew: 偏度
+ kurt: 峰度
+ quantile: 分位数

```python
data = pd.read_csv("./data/data9_5.txt", header=None)
data = data.T
print(data.describe())
print()
print(data.skew())           # 偏度
print()
print(data.kurt())           # 峰度
print()
print(data.quantile(0.9))    # 90% 分位数
```

#### 2.2.4 Q-Q 图

```python
from scipy.stats import norm

data = pd.read_csv("./data/data9_5.txt", header=None).T
data1 = data[0]
mu = data1.mean()
s = data1.std(ddof=1)
data1 = sorted(data1)
n = len(data1)
y = norm.ppf([(i - 0.5) / n for i in range(1, n + 1)], mu, s)
fig = go.Figure(data=[
    go.Scatter(x=y, y=data1, mode="markers", name="Q-Q 图"),
    go.Scatter(x=data1, y=data1, mode="lines", name="参照直线"),
])
fig.update_layout(width = 800, height = 800)
fig.update_xaxes(range=[20, 110])
fig.show()
```

## 3. 参数估计与假设检验

### 3.1 参数估计

```python
from scipy.stats import t, sem

data =np.loadtxt("./data/data9_11.txt").flatten()
n = len(data)
mu = data.mean()
s = data.std(ddof=1)
sem_value = sem(data)
print(f"SEM: {sem_value}")
alpha = 0.05
t_alpha = t.ppf(1 - alpha / 2, n - 1)
section = [mu - t_alpha * sem_value, mu + t_alpha * sem_value]
print(f"置信区间: {np.round(section, 4)}")
```

### 3.2 参数假设检验
直接套用概率论公式即可

Z 检验、t 检验等

### 3.3 非参数假设检验
分布拟合检验（$\chi^2$ 检验）

#### Kolmogorov-Smirnov 检验
原假设 $H_0: F(x) = F_0(x)$

检验统计量：$D_n=\sup\limits_{-\infty < x < \infty} |F_n(x)-F(x)|$

拒绝域：$D_n\geq \dfrac{t_\alpha}{\sqrt{n}}$

```python
from scipy.stats import kstest

data = []
with open("./data/data9_17.txt", 'r', encoding='utf-8') as f:
    for line in f:
        data.extend(line.split())
data = np.array(data, dtype=int)
x_bar = data.mean()
s = data.std(ddof=1)
dn, p_value = kstest(data, "norm", (x_bar, s))
print(f"D_n: {dn:.4f}, p-value: {p_value:.4f}")     # p 值 > 0.05 时，接受原假设
```

## 4. 方差分析

### 4.1 单因素方差分析

```python
import statsmodels.api as sm

data = pd.read_excel("./data/data9_18.xlsx", header=None)
data = data.values
y = data[~np.isnan(data)]
x = np.hstack([np.ones(5), np.full(4, 2), np.full(4, 3), np.full(3, 4)])
model = sm.formula.ols("y~C(x)", {"x": x, "y": y}).fit()        # C(x): 分类变量, 将 x 分为 4 个类别
anova = sm.stats.anova_lm(model)        # F: F 值, PR(>F): p 值
print(anova)        # p < 0.05, 拒绝原假设
```

### 4.2 双因素方差分析

```python
data = np.loadtxt("./data/data9_20.txt")
x1 = np.tile(np.arange(1, 5), (3, 1)).T
x2 = np.tile(np.arange(1, 4), (4, 1))
model = sm.formula.ols("y~C(x1)+C(x2)", {"x1": x1.flatten(), "x2": x2.flatten(), "y": data.flatten()}).fit()
anova = sm.stats.anova_lm(model)
print(anova)
```
